```python
from pathlib import Path
import shutil
import zipfile
import os

PROJECT_NAME = "Feature-Selection-Using-Metaheuristic-Approaches"

INPUT_ROOT = Path("/kaggle/input")
WORK_ROOT = Path("/kaggle/working")
PROJECT_DIR = WORK_ROOT / PROJECT_NAME

if not PROJECT_DIR.exists():
    # Kaggle usually extracts uploaded ZIP files automatically.
    candidates = []

    for requirements_file in INPUT_ROOT.rglob("requirements.txt"):
        candidate = requirements_file.parent

        if (
            (candidate / "notebooks").is_dir()
            and (candidate / "utils").is_dir()
            and (candidate / "datasets").is_dir()
        ):
            candidates.append(candidate)

    if candidates:
        source_dir = candidates[0]
        print("Copying from:", source_dir)

        shutil.copytree(source_dir, PROJECT_DIR)

    else:
        # Fallback in case Kaggle keeps the ZIP compressed.
        zip_files = list(
            INPUT_ROOT.rglob(
                "Feature-Selection-Metaheuristics-Kaggle-Ready.zip"
            )
        )

        if not zip_files:
            raise FileNotFoundError(
                "The uploaded project could not be found under /kaggle/input"
            )

        print("Extracting:", zip_files[0])

        with zipfile.ZipFile(zip_files[0]) as archive:
            archive.extractall(WORK_ROOT)

else:
    print("Reusing existing working project.")

if not PROJECT_DIR.exists():
    raise FileNotFoundError(
        f"Project folder was not created: {PROJECT_DIR}"
    )

os.chdir(PROJECT_DIR)

print("Current folder:", Path.cwd())
print("Project contents:")
print(sorted(path.name for path in PROJECT_DIR.iterdir()))
```

```python
from pathlib import Path
import subprocess

EXECUTED_DIR = Path("executed/smoke")
EXECUTED_DIR.mkdir(parents=True, exist_ok=True)


def run_notebook(filename):
    input_path = Path("notebooks") / filename
    output_name = filename.replace(
        ".ipynb",
        ".executed.ipynb",
    )

    print("=" * 70)
    print("Running:", filename)
    print("=" * 70)

    command = [
        "jupyter",
        "nbconvert",
        "--to",
        "notebook",
        "--execute",
        str(input_path),
        "--output",
        output_name,
        "--output-dir",
        str(EXECUTED_DIR),
        "--ExecutePreprocessor.timeout=-1",
    ]

    subprocess.run(command, check=True)

    print("Completed:", filename)
```

```python
from pathlib import Path
import shutil

WORK_DIR = Path("/kaggle/working")
PROJECT_NAME = "Feature-Selection-Using-Metaheuristic-Approaches"
ZIP_PATH = WORK_DIR / "Feature-Selection-Completed.zip"

if ZIP_PATH.exists():
    ZIP_PATH.unlink()

shutil.make_archive(
    str(WORK_DIR / "Feature-Selection-Completed"),
    "zip",
    root_dir=WORK_DIR,
    base_dir=PROJECT_NAME,
)

print("Fresh ZIP created:", ZIP_PATH)
print("Size:", round(ZIP_PATH.stat().st_size / 1024**2, 2), "MB")
```

```python
import zipfile

with zipfile.ZipFile(ZIP_PATH) as archive:
    final_files = [
        name for name in archive.namelist()
        if "/results/final/" in name
    ]

print("Final files inside ZIP:", len(final_files))

for name in final_files:
    print(name)

assert any(
    name.endswith("results/final/comparison_summary.csv")
    for name in final_files
)

print("\nZIP contains the final results.")
```
